```python
#bài 14
n=int(input())
lon_nhat=n
nho_nhat=n
while True:
    n=int(input())
    if n==-1:
        break
    if n>lon_nhat:
        lon_nhat=n
    if n<nho_nhat:
        nho_nhat=n
print(lon_nhat,nho_nhat,sep=' ')
```
