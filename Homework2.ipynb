```python
#bài 5
a,b,c=map(float,input().split())
if a==0:
    print('khong phai ptr bac 2')
else:
    delta=b**2-4*a*c
    if delta < 0:
        print('ptr vo nghiem')
    elif delta==0:
        print('ptr co nghiem kep:',round(-b/(2*a),2))
    else:
        x1= round((-b-delta**0.5)/(2*a),2)
        x2= round((-b+delta**0.5)/(2*a),2)
        print(f'ptr co 2 nghiem phan biet\nx1= {x1}\nx2= {x2}')
```
