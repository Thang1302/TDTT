### Problem 4
#### Write a function that returns the count of positive integers k such that 1 <= k < n and k > (n - 1) / 2. (Example: n=9 -> 4).

```python
def count_k(n):
    count=0
    for k in range(1,n):
        if k>(n-1)/2:
            count +=1
    return count
#print(count_k(200))
```

### Problem 6
#### Write a function that computes the area of a triangle with side lengths a, b, c (use Heron's formula). Raise ValueError if the sides cannot form a triangle.

```python
import math
def dien_tich(a,b,c):
    p=(a+b+c)/2
    if p*(p-a)*(p-b)*(p-c) <=0 or a<= 0 or b<=0 or c<= 0:
        return 'ValueError'
    s=math.sqrt(p*(p-a)*(p-b)*(p-c)) 
    return s
a,b,c=3,4,5
kq=dien_tich(a,b,c)
print(kq)



#cách khác

import math
def area(a,b,c):
    if a<=0 or b<=0 or c<=0 or a+b<=c or a+c<=b or b+c<=a:
        return 'ValueError'
    p=(a+b+c)/2
    heron = math.sqrt(p*(p-a)*(p-b)*(p-c))
    return heron
print(area(3,4,5))
```

### Problem 7
#### Write a function that returns True if a, b, c can be the side lengths of a triangle; otherwise return False.

```python
def canh_tam_giac(a,b,c):
    if a+b>c and b+c>a and a+c>b and a>0 and b>0 and c>0:
        return True
    else:
        return False
print(canh_tam_giac(1,2,5)) 

#cách 2 trả trực tiếp logic value
def tam_giac(a,b,c):
    return a>0 and b>0 and c>0 and a+b>0 and a+c>0 and b+c>0
print(tam_giac(3,4,5))
```
